# 서울시 교통 소외지역 GIS 분석: 250m 공간 격자 생성 및 시각화

본 노트북은 서울시 행정/법정동 경계 데이터로부터 **EPSG:5179 (ITRF2000 / UTMK)** 표준 좌표계를 기준으로 **250m x 250m 공간 격자(Grid)**를 생성하고, 이를 정적(Matplotlib) 및 대화형(Folium)으로 시각화합니다.

```python
import sys
from pathlib import Path

# Add project root directory to path
PROJECT_ROOT = Path("..").resolve()
if str(PROJECT_ROOT) not in sys.path:
    sys.path.append(str(PROJECT_ROOT))

import geopandas as gpd
from src import config
from src.data_loader import load_seoul_boundary, load_seoul_dongs
from src.grid_generator import generate_250m_grid
from src.visualization import plot_static_grid, plot_interactive_grid

print("Project Root:", PROJECT_ROOT)
print("Target CRS:", config.TARGET_CRS)
print("Grid Size:", config.GRID_SIZE_METERS, "meters")
```

## 1. 서울시 경계 데이터 로드 및 EPSG:5179 좌표 변환

```python
seoul_boundary = load_seoul_boundary()
print("Seoul Boundary CRS:", seoul_boundary.crs)
print("Seoul Total Area:", round(seoul_boundary.geometry.area.sum() / 1e6, 2), "km²")
seoul_boundary.head()
```

## 2. 250m 정방형 공간 격자망 생성

```python
seoul_grid = generate_250m_grid(seoul_boundary, grid_size=250)
print("Generated Grid Cell Count:", len(seoul_grid))
seoul_grid.head()
```

## 3. 정적 이미지 시각화 (Matplotlib)

```python
png_path = plot_static_grid(seoul_boundary, seoul_grid)
print("Static Map Saved to:", png_path)
```

## 4. 대화형 지도 시각화 (Folium)

```python
html_path = plot_interactive_grid(seoul_boundary, seoul_grid)
print("Interactive Map Saved to:", html_path)
```
